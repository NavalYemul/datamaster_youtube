## What is Databricks?

**Databricks** is a unified, open analytics platform for building, deploying, sharing, and maintaining enterprise-grade data, analytics, and AI solutions at scale. It integrates with cloud storage and security in your cloud account, and manages and deploys cloud infrastructure on your behalf.

### Core Components

| Component | Description |
|:---|:---|
| **Data Lakehouse** | Combines the best of data lakes and data warehouses — scalable storage with ACID transactions, schema enforcement, and BI performance |
| **Apache Spark** | Massively parallel compute engine for big data processing |
| **Delta Lake** | Open-source storage layer providing reliability, performance, and governance to data lakes |
| **Unity Catalog** | Unified governance solution for all data and AI assets |
| **Databricks SQL** | SQL-native analytics and warehousing experience |
| **MLflow** | Open-source platform for managing the end-to-end machine learning lifecycle |
| **Mosaic AI** | Tools for building and deploying generative AI and ML applications |

### Key Use Cases

- **Data Engineering** — Build reliable ETL/ELT pipelines with Delta Live Tables and Lakeflow
- **Data Warehousing & BI** — Run fast SQL analytics with serverless SQL warehouses
- **Data Science & ML** — Train, track, and deploy ML models at scale
- **Generative AI** — Build RAG apps, fine-tune LLMs, and serve foundation models
- **Real-Time Analytics** — Process streaming data for immediate insights

  <h2 style="color: #FF3621; margin: 0;">🏠 1. The Data Lakehouse</h2>

### What Is It?

A **data lakehouse** is an architecture that combines the best of **data lakes** (scalable, low-cost storage for all data types) with **data warehouses** (ACID transactions, schema enforcement, BI performance). Databricks builds the lakehouse on three pillars:

| Pillar | Role |
|:---|:---|
| **Apache Spark** | Massively scalable compute engine, decoupled from storage |
| **Delta Lake** | Optimized storage layer with ACID transactions, schema enforcement, and time travel |
| **Unity Catalog** | Unified, fine-grained governance for data and AI |

<hr style="border: none; height: 2px; background: linear-gradient(to right, #FF3621, #00A1E0, #86BC25); border-radius: 2px; margin: 20px 0;">

### Medallion Architecture (Bronze → Silver → Gold)

The recommended data design pattern that progressively improves data quality:

  <h2 style="color: white; margin: 0;">🐘 3. Lakebase Postgres</h2>
  <p style="margin: 5px 0 0 0; font-size: 14px;">A fully managed Postgres database integrated into the Databricks platform for real-time transactional workloads.</p>

### What Is It?

Lakebase is a **fully managed Postgres database** that lets you build real-time transactional (OLTP) applications alongside your lakehouse data. It features automatic scaling, instant branching, and native Unity Catalog integration.

<hr style="border: none; height: 2px; background: linear-gradient(to right, #FF3621, #00A1E0, #86BC25); border-radius: 2px; margin: 20px 0;">

### Architecture: Compute-Storage Separation

Lakebase separates stateless Postgres compute from a durable storage layer:

```
  ┌───────────────────────────────────────────────────────┐
  │  COMPUTE LAYER (Stateless Postgres)              │
  │  Runs the Postgres query engine                  │
  │  Autoscales • Scale-to-zero • Read replicas       │
  └───────────────────────────────────────────────────────┘
                         │
  ┌───────────────────────────────────────────────────────┐
  │  STORAGE LAYER (Durable, independent)             │
  │                                                   │
  │  📝 Safekeepers     WAL replication (Paxos quorum)│
  │  📄 Pageservers     On-demand page reconstruction  │
  │  ☁️  Object Storage  Persistent page storage        │
  └───────────────────────────────────────────────────────┘
```

| Traditional Monolith | Lakebase Component | Role |
|:---|:---|:---|
| Single machine | Stateless compute | Runs the Postgres query engine |
| Local WAL disk | **Safekeepers** | Durably records every committed change (Paxos quorum) |
| Local data files | **Pageservers + Object Storage** | Materializes and stores page versions on demand |

<div style="border-left: 5px solid #00A1E0; background: #e6f4fb; padding: 15px; margin: 15px 0; border-radius: 4px;">
  <strong>ℹ️ Note:</strong> Because data lives in cloud object storage rather than a single machine, Lakebase delivers elastic compute, durable cross-AZ writes, and no storage to provision -- you pay only for storage consumed.
</div>

<hr style="border: none; height: 2px; background: linear-gradient(to right, #FF3621, #00A1E0, #86BC25); border-radius: 2px; margin: 20px 0;">

### Key Features

| Feature | Description |
|:---|:---|
| ⚡ **Autoscaling** | Automatically adjust compute resources based on workload demand |
| 💤 **Scale to zero** | Suspend inactive computes to minimize costs |
| 🌿 **Branches** | Create isolated branches for dev and testing (instant, copy-on-write) |
| 📖 **Read replicas** | Scale read operations independently |
| ⏪ **Instant restore** | Create a new branch from any point in your history window |
| 🛡️ **High availability** | Automatic failover across availability zones |

<div style="padding: 15px; margin: 10px 0; background: #f9f9f9; border-left: 4px solid #1B3139; border-radius: 4px;">

<hr style="border: none; height: 2px; background: linear-gradient(to right, #FF3621, #00A1E0, #86BC25); border-radius: 2px; margin: 20px 0;">

### Security Model

| Aspect | Details |
|:---|:---|
| 🔐 **Identity** | Each app gets a dedicated, non-reusable service principal |
| 🛡️ **Permissions** | App SP needs explicit grants: CAN USE (warehouse), SELECT/MODIFY (tables), CAN QUERY (endpoints) |
| 👤 **User Authorization** | Optional: add API scopes so the app acts on behalf of the user (OAuth) |
| 💰 **Billing** | Per hour of compute time while running, based on provisioned capacity |

<div style="border-left: 5px solid #FF9100; background: #fff5e6; padding: 15px; margin: 15px 0; border-radius: 4px;">
  <strong>⚠️ Warning:</strong> Never hardcode personal access tokens in app code. Always use the auto-injected <code>DATABRICKS_CLIENT_ID</code> / <code>DATABRICKS_CLIENT_SECRET</code> environment variables. Apply <strong>least privilege</strong> -- grant only the minimum permissions the app needs.
</div>
