## 💻 1. Genie Code -- The Developer's AI Companion

> *The AI coding and data assistant embedded directly in the Databricks workspace.*

---

### 🎯 What It Does

Genie Code helps developers **generate, optimize, explain, and fix code** across notebooks, the SQL editor, Jobs, dashboards, and the file editor. It answers questions sourced from Databricks documentation and adapts to the product surface you're working in.

---

### ⭐ Core Features

| Feature | Description |
|:---|:---|
| 🤖 **Agent Mode** | Automates multi-step workflows: plans solutions, retrieves workspace assets, runs code, iterates on cell outputs, and fixes errors automatically. Adapts per surface (notebook = data exploration, pipeline editor = data engineering). |
| ⌨️ **AI Autocomplete** | Inline code completions as you type. |
| 💬 **Inline Chat** | Press `Cmd+I` / `Ctrl+I` in a code cell to ask questions or request code without leaving the cell. |
| 🔧 **Diagnose Error / Quick Fix** | Automatic error detection with one-click fix recommendations. |
| 🔍 **NL Data Filtering** | Filter DataFrames with plain English prompts. |
| 📅 **Scheduled Tasks** | Schedule Genie Code to run recurring tasks automatically and review each run. |
| 🌿 **Branch & Edit Chats** | Fork a chat from any point, use `/btw` for side questions, or edit an earlier message to regenerate. |

---

### 🧩 Extensibility

| Extension | Details |
|:---|:---|
| 📝 **Custom Instructions** | User-level and workspace-level instructions to guide response style, conventions, and preferences. |
| 🧰 **Agent Skills** | Reusable `SKILL.md` files that extend Genie Code with domain-specific knowledge for your team. |
| 🔌 **MCP Connectors** | Connect to external tools via Model Context Protocol: Google Drive, Gmail, Google Calendar, Microsoft 365, Atlassian, Glean, Slack, GitHub, or custom MCP servers. |

## 🧠 2. Genie Agents -- Domain-Specific Data Rooms

> *Conversational workspaces where data teams curate trusted datasets, business rules, and SQL logic for business users.*

---

### 🎯 What It Does

Genie Agents use a **compound AI system** to interpret business questions and generate governed answers. Instead of a single LLM, multiple interacting components parse the request, select context, generate read-only SQL, and return results with auto-visualizations.

---

### ⚙️ How a Response Is Generated

```
  💬 User Question
       │
       ▼
  ┌───────────────────────────────────────────────┐
  │  Context Assembly                          │
  │  ────────────────                          │
  │  • UC table metadata (names, PK/FK)        │
  │  • Column names & descriptions              │
  │  • Knowledge store snippets                 │
  │  • Example SQL queries                      │
  │  • Registered SQL functions                  │
  │  • General instructions                      │
  │  • Chat history                              │
  └───────────────────────────────────────────────┘
       │
       ▼
  📝 Generate Read-Only SQL
       │
       ▼
  ⚡ Execute on SQL Warehouse
       │
       ▼
  📊 Results + Auto-Visualization
```

---

### 🔀 Two Response Modes

| Mode | How It Works | Best For |
|:---|:---|:---|
| 💬 **Chat Mode** | Single-step: Genie picks the best table/query/function and returns one answer. Fast and simple. | Quick factual questions, KPI lookups |
| 🤖 **Agent Mode** *(formerly Research Agent)* | Multi-step reasoning: creates a research plan, runs multiple SQL queries, iterates on findings, delivers a comprehensive report with citations, visualizations, and supporting tables. Can also analyze unstructured files in UC Volumes. | Complex exploratory questions, trend analysis, root cause investigation |

---

### 🛠️ Author Configuration Toolkit

| Tool | Purpose |
|:---|:---|
| 📊 **Data Sources** | Attach specific UC tables; Genie only queries attached sources |
| 📝 **General Instructions** | Plain-text business rules, terminology, and guidance |
| 📎 **Example SQL Queries** | Verified queries Genie can match to user questions (with usage guidance) |
| ⚙️ **SQL Functions** | UC functions registered as trusted assets |
| 📖 **Knowledge Store** | Agent-local metadata: column values, JOIN relationships, SQL expressions for business semantics |
| 🎯 **Benchmarks** | Test questions with expected answers to score and monitor accuracy |
| 📊 **Quality Monitoring** | Review interactions, rate responses, request reviews from elevated-permission users |

---

### 🔗 APIs & Embedding

| API | Capability |
|:---|:---|
| **Chat Mode API** | Start conversations, send follow-ups, retrieve SQL + results + visualizations |
| **Agent Mode API** | Stream reasoning steps, SQL results, and final reports with citations |
| **Embeddable** | Embed a Genie Agent directly in external applications |

## 🌐 3. Genie One -- The Business User's Front Door

> *A simplified Databricks UI. One entry point for dashboards, data Q&A, and Apps -- no technical concepts required.*

---

### 🎯 What It Does

Genie One strips away workspace complexity (compute, queries, notebooks, models) and gives business users a clean interface to:

* 📊 **View AI/BI Dashboards** -- Browse and interact with shared dashboards to track KPIs and analyze metrics
* 💬 **Ask Data Questions** -- Unified full-screen NL chat that:
  * Searches available **Genie Agents** for a match first
  * Then searches across **dashboards, queries, and metric views**
  * Can connect **external sources** (Google Drive, SharePoint) for document-based Q&A
* 📱 **Use Databricks Apps** -- Custom-built apps combining analytics, AI, and workflows
* ✨ **Personalized Starter Questions** *(Beta)* -- One-click prompts tailored to your recent activity
* 📅 **Scheduled Recurring Chat Tasks** -- Set up tasks on a cadence that post results back as a thread

---

### 🌐 Two Access Levels

| Level | Scope | Who Can Access | How to Get There |
|:---|:---|:---|:---|
| 🏢 **Workspace-level** | Scoped to a single workspace | Workspace members with at least **Consumer Access** | Append `/one` to workspace URL, or use the app switcher |
| 🌍 **Account-level** | Unified view across all workspaces | All account users (see only assets shared with them) | `https://accounts.azuredatabricks.net/one` or workspace switcher |

> 💡 If a user's only entitlement is **Consumer Access**, they land on Genie One by default after login.

## 🏗️ Architecture -- How They Work Together

> *Three layers, one governed data platform.*

---

```
╔══════════════════════════════════════════════════════════════════════╗
║                                                                      ║
║   🌐  CONSUMPTION LAYER                                              ║
║   ┌──────────────────────────────────────────────────────────────┐   ║
║   │  GENIE ONE                                                    │   ║
║   │  Dashboards  |  NL Chat  |  Databricks Apps                  │   ║
║   │  👤 Business Users & Executives                               │   ║
║   └──────────────────────────────────────────────────────────────┘   ║
║                          │  routes questions to                     ║
║                          ▼                                          ║
║   🧠  CURATION LAYER                                               ║
║   ┌──────────────────────────────────────────────────────────────┐   ║
║   │  GENIE AGENTS                                                 │   ║
║   │  Curated Tables | Instructions | Trusted SQL | Knowledge     │   ║
║   │  👤 Data Analysts & Agent Authors                             │   ║
║   └──────────────────────────────────────────────────────────────┘   ║
║                          │  queries governed by                     ║
║                          ▼                                          ║
║   💻  DEVELOPMENT LAYER                                             ║
║   ┌──────────────────────────────────────────────────────────────┐   ║
║   │  GENIE CODE                                                   │   ║
║   │  Pipelines | Models | Notebooks | Dashboards | Data          │   ║
║   │  👤 Developers, Data Engineers & Data Scientists               │   ║
║   └──────────────────────────────────────────────────────────────┘   ║
║                          │  all governed by                         ║
║                          ▼                                          ║
║            ┌──────────────────────────────────────────┐        ║
║            │    🛡️  UNITY CATALOG  (Governance)       │        ║
║            │    Permissions | Lineage | Metadata       │        ║
║            └──────────────────────────────────────────┘        ║
║                                                                      ║
╚══════════════════════════════════════════════════════════════════════╝
```

---

### 🔄 The Flow

| Step | Who | What |
|:---:|:---|:---|
| **1** | 👩‍💻 **Developers** | Use **Genie Code** to build data pipelines, models, dashboards, and curate tables in Unity Catalog |
| **2** | 👩‍💼 **Analysts** | Create **Genie Agents** on top of those tables -- adding business rules, instructions, trusted SQL, and quality benchmarks |
| **3** | 👤 **Business Users** | Access **Genie One** to ask questions (routed to Genie Agents), browse dashboards, and use Apps |

---
