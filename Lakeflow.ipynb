## 📌 What is Lakeflow?
Lakeflow is Databricks' **unified data engineering solution** — ingestion, transformation, and orchestration on one platform, governed by Unity Catalog.

## 🧩 The 4 Core Components
| Component | Purpose | Use When |
|---|---|---|
| **Lakeflow Connect** | Managed ingestion connectors | Data lives outside Databricks |
| **Lakeflow Designer** | No-code visual pipelines (Public Preview) | Analysts, drag-and-drop ETL |
| **Lakeflow Spark Declarative Pipelines** | Pro-code ETL (formerly DLT) | SQL/Python, streaming + batch |
| **Lakeflow Jobs** | Orchestration (formerly Workflows) | Scheduling & running everything |
