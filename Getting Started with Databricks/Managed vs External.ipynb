| **Feature**        | **Managed Table**                                       | **External Table**                                      |
|--------------------|------------------------------------------------|------------------------------------------------|
| **Create Table Syntax** | `CREATE TABLE table_name (...)`  | `CREATE TABLE table_name LOCATION 'path' (...)` |
| **Format**        | Delta                                          | Any format (Delta, CSV, Parquet, etc.)         |
| **Location**      | Default storage location (Unity Catalog) | Must be explicitly specified (ADLS, S3, etc.)  |
| **Drop Behavior** | Data and metadata are deleted                  | Only table is removed from Unity Catalog, data remains |
| **Undrop**        | Supported (`UNDROP TABLE` command available)  | Not supported                                   |
| **Optimization**  | Predictive optimization supported              | Manual    |
