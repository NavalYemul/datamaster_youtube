## 🤖 What Are Databricks AI Functions?

**AI Functions** are built-in Databricks functions that allow you to apply **Artificial Intelligence** capabilities—such as **text translation** or **sentiment analysis**—directly on data stored in Databricks.

These functions can be executed from **anywhere within the Databricks platform**, including:

- Databricks SQL  
- Notebooks  
- Lakeflow Spark Declarative Pipelines  
- Workflows  

AI Functions make it easy to enrich your data with **AI-powered insights** without building, training, or managing machine learning models.

#### 🤖 `ai_analyze_sentiment()`

The **`ai_analyze_sentiment()`** function allows you to invoke a **state-of-the-art generative AI model** to perform **sentiment analysis** on input text using **SQL**.

It analyzes textual data and determines whether the sentiment expressed is:
- **Positive**
- **Negative**
- **Neutral**
- **Mixed**

This function enables teams to apply **AI-driven sentiment insights directly on structured or unstructured text data**—without building, training, or deploying machine learning models.


```sql
SELECT ai_analyze_sentiment('I am happy')
```

```sql
SELECT ai_analyze_sentiment('I am sad');
```

#### 🤖 `ai_classify()`

The **`ai_classify()`** function allows you to use a **state-of-the-art generative AI model** to classify input text into predefined categories using **SQL**.

You provide:
- The input text
- A list of possible categories

The function then analyzes the text and assigns it to the most relevant category.

This is useful for tasks such as:
- Ticket categorization (Billing, Technical, General Inquiry)
- Email routing
- Product review classification
- Intent detection

With `ai_classify()`, you can add **AI-powered classification directly to your data pipelines**—without building or training a machine learning model.

##### Syntax
ai_classify(content, labels)

```sql
SELECT ai_classify("My password is leaked.", ARRAY("urgent", "not urgent"))
```

```sql
SELECT
    description,
    ai_classify(description, ARRAY('clothing', 'shoes', 'accessories', 'furniture')) AS category
  FROM
    products
  LIMIT 10;
```

```sql
--Classification (ticket routing)
SELECT
  ticket_id,
  ticket_subject,
  ai_classify(
    ticket_text,
    ARRAY('billing', 'delivery', 'technical', 'account', 'other')
  ) AS ticket_category
FROM support_tickets;
```

#### 🤖 `ai_extract()`

The **`ai_extract()`** function allows you to invoke a **state-of-the-art generative AI model** to extract structured information from unstructured text using **SQL**.

It identifies and extracts specific entities or fields from text data, returning structured outputs based on your defined schema.

This is useful for tasks such as:
- Extracting names, dates, and locations from documents
- Pulling invoice details from emails
- Extracting order numbers from support tickets
- Converting free-text into structured columns

With `ai_extract()`, you can transform unstructured text into **structured, queryable data**—without writing complex NLP pipelines.

```sql
> SELECT ai_extract(
    'John Doe lives in New York and works for Acme Corp.',
    array('person', 'location', 'organization')
  );
 {"person": "John Doe", "location": "New York", "organization": "Acme Corp."}

> SELECT ai_extract(
    'Send an email to jane.doe@example.com about the meeting at 10am.',
    array('email', 'time')
  );
 {"email": "jane.doe@example.com", "time": "10am"}
```

```sql
select * from customer_messages
```

```sql
--Entity extraction (order ID, city, etc.)
SELECT
  message_id,
  message_text,
  ai_extract(
    message_text,
    ARRAY('credit_card_number', 'phone_number', 'email', 'address', 'order_id', 'city')
  ) AS extracted_entities
FROM customer_messages;
```

#### 🤖 `ai_gen()`

The **`ai_gen()`** function allows you to invoke a state-of-the-art generative AI model to generate text based on a given prompt using SQL.

You can dynamically combine:
- Static instructions (prompt)
- Column values from your tables

The function generates contextual AI responses directly within your query results.

Common use cases include:
- Generating customer support replies
- Creating product descriptions
- Summarizing support tickets
- Drafting email responses
- Producing FAQ answers

With `ai_gen()`, you can integrate generative AI directly into your data workflows—without managing external APIs or deploying language models.

```sql
SELECT ai_gen('Generate a concise, cheerful email title for a summer bike sale with 20% discount');
  Summer Bike Sale: Grab Your Dream Bike at 20% Off!
```

```sql
--Text generation (response suggestion)
SELECT
  ticket_id,
  ticket_subject,
  ai_gen(
    CONCAT(
      'You are a polite customer support agent. Write a short reply (3–4 sentences) to this customer ticket:\n\n',
      ticket_text
    )
  ) AS reply_suggestion
FROM support_tickets;
```

#### ✍️ `ai_fix_grammar()`

The **`ai_fix_grammar()`** function allows you to use a state-of-the-art generative AI model to automatically correct grammar and improve the clarity of input text using SQL.

It refines:
- Spelling mistakes
- Grammatical errors
- Sentence structure
- Writing clarity

This function is useful for:
- Cleaning customer messages
- Improving support responses
- Standardizing communication across teams
- Enhancing text quality before analysis

With `ai_fix_grammar()`, you can automatically improve text quality at scale directly within Databricks.


```sql
> SELECT ai_fix_grammar('This sentence have some mistake');
 "This sentence has some mistakes"

> SELECT ai_fix_grammar('She dont know what to did.');
 "She doesn't know what to do."

> SELECT ai_fix_grammar('He go to school every days.');
 "He goes to school every day."
```

```sql
--Grammar correction
SELECT
  ticket_id,
  ticket_text AS original_text,
  ai_fix_grammar(ticket_text) AS corrected_text
FROM support_tickets;
```

```sql
--Mask PII
SELECT
  message_id,
  message_text AS original_text,
  ai_mask(
    message_text,
    ARRAY('credit_card_number', 'phone_number', 'email', 'address', 'pan')
  ) AS masked_text
FROM customer_messages;
```

```sql
--Summarization
SELECT
  review_id,
  product_name,
  ai_summarize(review_text) AS short_summary
FROM customer_reviews;
```

```sql
--Translation (Hindi → English demo)

SELECT
  review_id,
  review_text AS original_text,
  ai_translate(
    review_text,
    'en'
  ) AS translated_to_english
FROM customer_reviews
WHERE language = 'hi';
```
